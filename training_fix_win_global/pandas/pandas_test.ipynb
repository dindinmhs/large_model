#### Task 5: Label Encoding (Competition Style - Medium)
Create mapping dictionaries to convert categorical text features into integer arrays for the model. 
Map the `Sex` column (`'male'` to `1`, `'female'` to `0`) and the `Embarked` column (`'S'` to `0`, `'C'` to `1`, `'Q'` to `2`). Apply these mappings to modify the columns in the DataFrame directly.

#### Task 6: String Manipulation & Text Truncation (Competition Style - Medium)
Using Pandas string accessors, create a new column named `Short_Name` that takes the original `Name` column, converts it entirely to lowercase, splits the text into words, and keeps **only the first two words** of the name. Join them back as a string.

```python
# Write your code here
def process(text:str):
    text = text.lower()
    text = text.split()
    text = text[:2]
    text = ' '.join(text)
    return text
# df['Short_Name'] = df['Name'].str.lower().str.split().str[:2].str.join(' ')
df['Short_Name'] = df['Name'].apply(process)
df
```

#### Task 7: Text Extraction & Feature Engineering (Medium)
Extract the passenger's title (e.g., 'Mr', 'Mrs', 'Miss', 'Master') from the `Name` column using regular expressions and store it in a new column called `Title`. Then, replace all rarely occurring titles (anything other than Mr, Mrs, Miss, Master) with a single string value `'Rare'`.

```python
# Write your code here
df['Title'] = df['Name'].str.extract(r' ([A-Za-z]+)\.')
main_titles = ['Mr', 'Mrs', 'Miss', 'Master']
df['Title'] = df['Title'].apply(lambda x: x if x in main_titles else 'Rare')
df
```

#### Task 8: Multi-level Aggregation (Medium)

#### Task 10: Contextual Missing Value Imputation (Medium-Hard)
Fill the missing values in the `Age` column by using the median age of passengers sharing the exact same `Pclass` and `Sex`. After that, drop any rows that have missing values in the `Embarked` column.

```python
# Write your code here
print(df.isna().sum())
median_age = df["Age"].median()
df['Age'] = df['Age'].fillna(df.groupby(['Pclass', 'Sex'])['Age'].transform('median'))
df = df.dropna(subset=['Embarked'])
print(df.isna().sum())
```

#### Task 12: Class Balancing & Downsampling (Competition Style - Hard)
Determine the number of passengers who survived and who did not survive. Find the minimum count between these two classes. Then, downsample the majority class so that the DataFrame contains an exactly equal number of rows for both `Survived = 0` and `Survived = 1`.

```python
# Write your code here
count = df.groupby('Survived').count()
min = count.min().iloc[0]
min

df = df.drop('Cabin', axis='columns')
df = df.groupby('Survived', group_keys=False).apply(lambda x : x.sample(n=min, random_state=42))
df.groupby('Survived').count()
```
