```python
def preprocess_text(text:str, max_lenght:int=150):
    text = text.split()
    text = text[:max_lenght]
    text = ' '.join(text)
    text = text.lower()
    return text
```

```python
df['text'] = df['text'].apply(lambda x : preprocess_text(x))
df['text']
```

```python
df['word_count'] = df['text'].str.split().str.len()
df
```

```python
sentiment_map = {0 : 'negative', 2 : 'neutral', 4 : 'positive'}
df['sentiment_label'] = df['target'].map(sentiment_map)
df[['sentiment_label', 'target']]
```
