```python
from datasets import load_dataset
import pandas as pd
import os

print("Mengunduh dataset Alpaca...")
ds = load_dataset("tatsu-lab/alpaca", split="train")

df = ds.to_pandas()

output_dir = "data"
os.makedirs(output_dir, exist_ok=True)
csv_path = os.path.join(output_dir, "alpaca.csv")

df.to_csv(csv_path, index=False)

print(f"Selesai! Dataset berhasil disimpan di: {csv_path}")
print(f"Total baris data: {len(df)}")
```
